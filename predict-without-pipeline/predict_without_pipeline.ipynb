```python
import pickle
import numpy as np
```

```python
ohe_embarked=pickle.load(open('ohe_embarked.pkl', 'rb'))
ohe_sex=pickle.load(open('ohe_sex.pkl', 'rb'))
clf=pickle.load(open('clf.pkl', 'rb'))
```

```python
test_input=np.array([2, 'male', 31.0, 0, 10.5, 50.0, 'S'], dtype=object).reshape(1,7)
```

```python
test_input
```

```python
test_input_sex=ohe_sex.transform(test_input[:,1].reshape(1,1))
```

```python
test_input_sex
```

```python
test_input_embarked=ohe_embarked.transform(test_input[:,-1].reshape(1,1))
test_input_embarked
```

```python
test_input_age=test_input[:,2].reshape(1,1)
test_input_age
```

```python
rmed=np.concatenate((test_input[:,[0,3,4,5]], test_input_age, test_input_sex, test_input_embarked), axis=1)
```

```python
rmed.shape
```

```python
clf.predict(rmed)
```
